# 🔹 Inheritance kya hota hai? (Core idea)

    Inheritance =

    Ek class (Child) dusri class (Parent) ke properties + methods use karti hai

    👉 Code reuse
    👉 Real-world hierarchy

## Real-life example 👇

    Animal → Dog → Puppy

    Employee → Manager

    Vehicle → Car → ElectricCar

````python
class Parent:
    def show(self):
        print('I am Parent')

class Child(Parent):
    pass

c = Child()
c.show()
# ✔ Child ne Parent ka method use kar liya
````

## TYPES OF INHERITANCE
### 1️⃣ Single Inheritance

````python
class Animal:
    def eat(self):
        return 'Eating'

class Dog(Animal):
    def bark(self):
        return 'Barking'
        
# ✔ Simple parent → child
      
dog = Dog()
print(dog.bark())
print(dog.eat())
````

### 2️⃣ Multilevel Inheritance

````python
class Animal:
    def eat(self):
        return 'Eating'

class Dog(Animal):
    def bark(self):
        return 'Barking'

class Puppy(Dog):
    def play(self):
        return 'Playing'
    
p = Puppy()
print(p.eat())
print(p.bark())
print(p.play())
````

### 3️⃣ Hierarchical Inheritance

````python
class Animal:
    def eat(self):
        return 'Eating'
    
class Dog(Animal):
    pass

class Cat(Animal):
    pass
# ✔ Ek parent → multiple children
````

### 4️⃣ Multiple Inheritance 🔥 (Confusing but powerful)

````python
class Father:
    def skill(self):
        return 'Driving'

class Mother:
    def skill(self):
        return 'Cooking'

class Child(Father, Mother):
    pass

c = Child()
print(c.skill())
````

    ❓ Output kya hoga?

    👉 Father ka method
    Kyuki left-to-right order

### 🔥 METHOD RESOLUTION ORDER (MRO) 🧠
```
Python ka rule:
Method pehle Child → Parent (left to right) dhundhta hai

````python
print(Child.mro())
````

    👉 object sabka ultimate parent hota hai

## 🔹 Interview Favorite ❗

````python
class A:
    def show(self):
        return 'A'
    
class B(A):
    def show(self):
        return "B"

class C(A):
    def show(self):
        return 'C'

class D(B, C):
    pass

d = D()
print(d.show())
print(D.mro())
````

    👉 Python C3 Linearization use karta hai
    👉 Isliye ambiguity nahi hoti

## 🔥 super() keyword (VERY IMPORTANT)
    ❓ super() kya karta hai?

    Parent class ka method call karta hai (MRO follow karke)
### 🔹 Without super() (BAD practice)

````python
class Parent:
    def __init__(self):
        return 'Parent init'
    

class Child(Parent):
    def __init__(self):
        Parent.__init__(self)
        return 'Child init'
````

## 🔹 With super() (GOOD practice)

````python
class Parent:
    def __init__(self):
        return 'Parent init'
    
class Child(Parent):
    def __init__(self):
        super().__init__(self)
        return 'Child init'
# ✔ Cleaner
# ✔ Multiple inheritance safe
````

### 🔥 Method Overriding 🎭

    Same method name, child ka method chalega

````python
class Animal:
    def sound(self):
        return 'Animal Sound'

class Dog(Animal):
    def  sound(self):
        return 'Bark'
d = Dog()
d.sound()
````

### 🔹 Parent method bhi chahiye? Use super()


````python
class Animal:
    def sound(self):
        return 'Animal Sound'
    
class Dog(Animal):
    def sound(self):
        perent_sound = super().sound()
        return f"{perent_sound} Bark"
    
d = Dog()
d.sound()
````

# 📝 PRACTICE TASK (VERY IMPORTANT)

````python
class Person:
    def __init__(self, name):
        self.name = name
    
    def show(self):
        return f"Name: {self.name}"
    
    
class Employee(Person):
    def __init__(self, name, salary):
        parent_data = super().__init__(name)
        self.salary = salary
        
    def show(self):
        parent_data = super().show()
        return f"{parent_data} ,Salary: {self.salary}"
    
# ✔ Object banao
# ✔ show() call karo
# ✔ MRO print karo


e = Employee('Harsh', 20000)
print(e.show())
print(Employee.mro())
````
