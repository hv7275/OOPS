### 🔥 TYPES OF POLYMORPHISM IN PYTHON

#### Python mein mainly 4 ways se polymorphism hota hai:

    1. Method Overriding

    2. Duck Typing

    3. Operator Overloading

    4. Function Polymorphism

### 1️⃣ Method Overriding (Inheritance ke saath)
    Child class parent ke method ko replace kar deta hai.

```python
class Shape:
    def area(self):
        return 'Area is not defined'

class Ractangle(Shape):
    def area(self):
        return f"Area = length * breadth"
    
s = Ractangle()
s.area()
```

    👉 Runtime pe decide hota hai
    👉 Runtime Polymorphism
#### 🔹 Rule yaad rakho ❗

    Method name same

    Parameters same

    Child ka method chalega

### 2️⃣ Duck Typing 🦆 (Python ka SUPERPOWER)

    “If it looks like a duck and quacks like a duck, it is a duck”

    Python ko class matter nahi karti,
    Python ko method matter karta hai.

### Example

```python
class Dog:
    def speak(self):
        return "Bark"

class Human:
    def speak(self):
        return 'Hello'
```

```python
def make_sound(obj):
    return obj.speak()
print(make_sound(Dog()))
print(make_sound(Human()))
```

    ✔ No inheritance
    ✔ Same method name
    ✔ Python happy 😎
#### 🔥 Interview Line:

    Python follows duck typing, not strict type checking

### 3️⃣ Operator Overloading 🔢

    Same operator, different behavior.

```python
print(10 + 20)
print("Hello" + " World")
```

### Custom class example

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

p1 = Point(1, 2)
p2 = Point(3, 4)

p3 = p1 + p2
print((p3.x, p3.y))
```

    👉 + internally calls __add__()
| Operator | Method    |
| -------- | --------- |
| +        | `__add__` |
| -        | `__sub__` |
| *        | `__mul__` |
| ==       | `__eq__`  |
| <        | `__lt__`  |


### 4️⃣ Function Polymorphism

    Same function, different types.

```python
print(len("Python"))
print(len([1, 2, 3, 4]))
print(len({1, 2, 3}))
```

    👉 Same function name
    👉 Different data types
    👉 Python handles it

### 🔥 Method Overloading in Python? ❌

```python
class Test:
    def add(self, a, b):
        return a + b
    
    def add(self, a, b, c):
        return a + b + c
```

    ❌ Second method overwrite kar dega first ko
### ✅ Python alternative (default args)

```python
class Test:
    def add(self, a, b=0, c=0):
        return a + b + c
    
t = Test()
t.add(5, 5)
```

## 🔥 REAL-WORLD EXAMPLE (Polymorphism)

```python
class Payment:
    def pay(self):
        pass
    
class UPI:
    def pay(self):
        return 'Paying via UPI'

class Card:
    def pay(self):
        return 'Paying via Card'

def make_payment(payment):
    return payment.pay()

print(make_payment(UPI()))
print(make_payment(Card()))
```

    ✔ Same interface
    ✔ Different implementation
    ✔ Clean design

    🔥 INTERVIEW QUESTIONS
    Q1. Compile-time vs Runtime polymorphism?

    ➡️ Python = mostly Runtime

    Q2. Duck typing kya hota hai?

    ➡️ Behavior-based polymorphism

    Q3. Method overloading supported?

    ➡️ ❌ Not directly

```python
class File:
    def read(self):
        return 'Reading file'

class Audio:
    def read(self):
        return 'Playing Audio'
    
class Video:
    def read(self):
        return 'Playing Video'

def play(media):
    return media.read()

f = File()
a = Audio()
v = Video()

print(play(f))  # Reading file
print(play(a))  # Playing Audio
print(play(v))  # Playing Video
```
