# Foundation

````python
# Class ek template hoti hai jisse objects bante hain.

'''
__init__() kya hai?

Isse constructor bolte hain

Object banate time automatically call hota hai

self → current object ko refer karta hai

👉 self bina likhe bhi object ke variables access nahi hote
'''

class Student:
    def __init__(self, name, age):
        self.age = age
        self.name = name
    
    def introduce(self):
        print(f"My name is {self.name} and age is {self.age}")
````

````python
s1 = Student('Harsh', 20)
s2 = Student('Khushi', 20)

print(s1.introduce())
print(s2.introduce())
````

# 2️⃣ Encapsulation (Data ko protect karna 🔐)

````python
'''

Data + Methods ko ek jagah band karna aur direct access restrict karna

🔹 Access Modifiers (Python style)
Type	Meaning
Public	Normal variable
Protected	_variable
Private	__variable

🔥 Real Life Example

ATM machine → tum directly bank balance change nahi kar sakte
Sirf methods (deposit, withdraw) ke through hota hai'''

class BankAccount:
    def __init__(self, amount):
        self.__amount = amount   # private variable
        
    def deposit(self, amount):
        self.__amount += amount  # controlled access
        
    def get_balance(self):
        return self.__amount     # getter

acc = BankAccount(10000)
acc.deposit(5000)
print(acc.get_balance())
````

# 3️⃣ Inheritance (Parent → Child 👨‍👦)

Ek class dusri class ki properties use kare = Inheritance

````python
class Animal:
    def speak(slef):
        print('Animal speak')

class Dog(Animal):
    def bark(slef):
        print('Dog Bark')

d = Dog()
d.speak()   # Parent method
d.bark()    # Child method
````

## 🔹 Types of Inheritance
```
1. Single

2. Multiple

3. Multilevel

4. Hierarchical
```

### 🔹 Multiple Inheritance

````python
class Father:
    def skill1(self):
        print('Driving')

class Mother:
    def skill2(self):
        print('Cooking')

class Child(Father, Mother):
    pass
````

# 4️⃣ Polymorphism (Same naam, different kaam 🎭)
```Polymorphism = Many forms```

````python
'''
➡️ Same method name, different behavior
'''
class Bird:
    def fly(self):
        print('Bird can Fly')
    
class Penguin(Bird):
    def fly(self):
        print('Penguin cannot fly')
        
b = Penguin()
b.fly()
````

````python
print(5 + 3)
print("Hello " + "World")
# Same + operator, alag-alag kaam 😎
````

# 5️⃣ Abstraction (Sirf zaroori cheez dikhao 👀)

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
    
class Car(Vehicle):
    def start(self):
        print('Car start with key')

c = Car()
c.start()
# ❌ Direct object nahi bana sakte:
````

Car has-a Engine → Composition

### 🔹 __str__() magic method

````python
class Book:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name
b = Book("Python OOP")
print(b)
````
