## 🔥 Abstraction vs Encapsulation (Confusion Clear)
| Encapsulation     | Abstraction      |
| ----------------- | ---------------- |
| Data hide         | Logic hide       |
| Security focus    | Design focus     |
| Private variables | Abstract classes |

    Encapsulation = “Data protect karo”
    Abstraction = “User ko sirf interface dikhao”

## 🔷 Python mein Abstraction kaise hota hai?

    Python mein Abstract Base Class (ABC) use karte hain.

    Module:

```python
from abc import ABC, abstractmethod
```

### 🔹 Basic Abstract Class Example

```python
class Vehicle(ABC):
    
    @abstractmethod
    def start(self):
        pass

v = Vehicle()  # ❌ ERROR
```

    Abstract class ka object nahi bana sakte.
### 🔹 Child Class Implementation

```python
class Car(Vehicle):
    def start(self):
        return  'Car Start With Key'
c= Car()
c.start()
```

    🔥 Rule yaad rakho ❗

    Agar child class abstract method implement nahi kare:

```python
class Bike(Vehicle):
    pass
b = Bike()  # ❌ ERROR
```

    Python bolega:

    Can't instantiate abstract class

## 🔥 Real-World Design Example

    Socho tum ek Payment System bana rahe ho.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    
    
    @abstractmethod
    def pay(self, amount):
        pass
   
# UPI Implementation 
class UPI(Payment):
    def pay(self, amount):
        return f'Paid {amount} using UPI'
    
# Card Implementation
class  Card(Payment):
    def pay(self, amount):
        return f'Paid {amount} using Card'
    
def process_payment(payment: Payment):
    return payment.pay(1000)

print(process_payment(UPI()))
print(process_payment(Card()))
```

    👉 Ye hai clean architecture
    👉 Interface fixed hai
    👉 Implementation flexible hai

### 🔥 Abstract Class mein Normal Methods bhi ho sakte hain

```python
class Shape(ABC):
    def info(self):
        return 'This is a Shape'
    
    @abstractmethod
    def area(self):
        pass
```

    ✔ Abstract class mein normal methods allowed
    ✔ Concrete methods + abstract methods dono ho sakte hain

### 🔥 When to Use Abstraction?

    Use abstraction when:

    ✔ Tum framework bana rahe ho
    ✔ Multiple implementations possible ho
    ✔ Team project mein contract define karna ho
    ✔ Large scale backend systems
