# ENCAPSULATION
## 🔹 Encapsulation kya hota hai? (Real meaning)

    Encapsulation =

    Data (variables) + Methods (functions) ko ek unit (class) mein band karna
    aur direct access ko control karna

### 💡 Goal:

    Data safe rahe

    Galat usage na ho

    Controlled access mile

### 🔹 Real-life example 🏦

    ATM Machine

    Tum directly balance change nahi kar sakte ❌

    Sirf methods: withdraw(), deposit() ✔

    Balance hidden hota hai

    👉 Yahi Encapsulation hai.

### 🔹 Python mein Encapsulation ka truth 😈

    ⚠️ Important:
    Python mein true private variables nahi hote (like Java / C++)

    👉 Python “consenting adults” philosophy follow karta hai
    (matlab: programmer samajhdaar hai 😄)

| Type      | Syntax | Meaning                  |
| --------- | ------ | ------------------------ |
| Public    | `x`    | Koi restriction nahi     |
| Protected | `_x`   | “Please access mat karo” |
| Private   | `__x`  | Name mangling hota hai   |


### Public Member

````python
class Student:
    def __init__(self, name):
        self.name = name
s = Student("Rahul")
print(s.name)   # Allowed
````

### 🔹 2️⃣ Protected Members (_variable)

````python
class Student:
    def __init__(self, marks):
        self._marks = marks

s = Student(90)
print(s._marks)   # Allowed but NOT recommended
# 👉 Yeh sirf convention hai, security nahi.
````

### 🔹 3️⃣ Private Members (__variable) 🔐

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
acc = BankAccount(1000)
print(acc.balance)   # ❌ ERROR
````

#### 🔥 Name Mangling (MOST IMPORTANT)

```

print(acc._BankAccount__balance)  # 😈

### 🔹 Data ko safely access kaise kare? (GETTER & SETTER)
```
❌ Bad practice
acc.__balance = -500
```
### ✅ Correct way

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    
    def get_balance(self):
        return self.__balance
    
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance += amount
        else:
            print('Invailid Amount')

acc = BankAccount(1000)
acc.set_balance(5000)
print(acc.get_balance())
````

### 🔹 Pythonic Way: @property (VERY IMPORTANT 🔥)

    Yeh industry standard hai.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
        
    @property
    def balance(self):
        return self.__balance   # ✅ return the private attribute
    
    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount   # ✅ set, don’t add
        else:
            raise ValueError('Invalid Amount')

acc = BankAccount(1000)
print(acc.balance)   # 1000

acc.balance = 2000
print(acc.balance)   # 2000
````

    👉 Actually method hai 😎

# 🔥 Interview Questions (Encapsulation)
    Q1. Python mein private variable kaise kaam karta hai?

    ➡️ Name mangling ke through

    Q2. _x aur __x mein difference?

    ➡️ Convention vs name mangling

    Q3. Kya Python secure hai?

    ➡️ Control ke liye hai, force ke liye nahi

````python
class Employee:
    def __init__(self, salary):
        self.__salary = salary
    
    @property
    def salary(self):
        return self.__salary
    
    @salary.setter
    def salary(self, value):
        if value >= 0:
            self.__salary = value
        else:
            raise ValueError('Invalid Value')

e = Employee(1000)
print(e.salary)

e.salary = 20000
print(e.salary)
# ✔ Salary negative na ho
# ✔ Read & write both allowed
# ✔ Error raise karo
````
