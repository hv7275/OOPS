# 🔷 1️⃣ Instance Method (Most Common)
    🔹 Definition

Jo method object ke data (instance variables) ke saath kaam kare.

👉 Pehla parameter self hota hi hota hai

### ❓ Kab use kare?

    ✔ Jab object-specific data use karna ho
    ✔ Jab behavior object pe depend kare

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
        
    def show(self):
        return f"{self.name}, {self.marks}"
    
s1 = Student("Rahul", 85)
s2 = Student("Anita", 92)

print(s1.show())
print(s2.show())
````

## 🔷 2️⃣ Class Method (@classmethod)
    🔹 Definition

Jo method class ke data (class variables) ke saath kaam kare.

👉 Pehla parameter = cls (class ka reference)

````python
class Student:
    college = 'IIT'
    
    def __init__(self, name):
        self.name= name
        
    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college

s1 = Student("Rahul")
s2 = Student("Anita")

Student.change_college("NIT")

print(s1.college)
print(s2.college)
````

## `Student.change_college("NIT")` vs `s1.change_college("NIT")`

Both of these lines **work**, but they are **not equal in meaning or intent**.  
Python allows both — **best practice decides which one you should use**.

---

## Example Code

```python
class Student:
    college = "IIT"

    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college

```
### Explanation
    college is shared by all students

    The change affects every object

    Calling via Student clearly shows global impact


## 🔥 Factory Method Example (Interview Favorite)

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        
    @classmethod
    def from_string(cls, data):
        name, salary = data.split(',')
        return cls(name, int(salary))
e = Employee.from_string("Amit,50000")
print(e.name, e.salary)
````

## 🔷 3️⃣ Static Method (@staticmethod)
#    🔹 Definition
    Jo method:

1. Na object ka use kare

2. Na class ka use kare

3. Bas logically related ho class se.

```
👉 No self, No cls

❓ Kab use kare?

✔ Utility / helper functions
✔ Logical grouping ke liye

````python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b
print(MathUtils.add(10, 20))
````

| Feature              | Instance | Class          | Static          |
| -------------------- | -------- | -------------- | --------------- |
| Decorator            | ❌        | `@classmethod` | `@staticmethod` |
| First param          | `self`   | `cls`          | ❌               |
| Access instance vars | ✔        | ❌              | ❌               |
| Access class vars    | ✔        | ✔              | ❌               |
| Object required      | ✔        | ❌              | ❌               |

## 🧠 MEMORY & DESIGN THINKING

    Instance method → behavior belongs to object

    Class method → behavior belongs to class

    Static method → behavior belongs to concept

## 🔥 INTERVIEW TRAPS ❗
    ❓ Can static method access class variable?

    ❌ Directly nahi
    ✔ Agar class name se access karo toh possible


````python
class Test:
    x = 10
    
    @staticmethod
    def show():
        print(Test.x)

t= Test()
t.show
````

### ❓ Can we overload methods in Python?

    ❌ Nahi (same name methods overwrite ho jaate hain)

````python
class Bank:
    bank_name = 'SBI'
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    
    def deposit(self, amount):
        if Bank.is_valid_amount(amount):
            self.balance += amount
            print(f"₹{amount} deposited successfully")
        else:
            print('Invailid Amount')
    
    @classmethod
    def change_bank(cls, name):
        cls.bank_name = name
    
    @staticmethod
    def is_valid_amount(amount):
        return amount > 0
    
b1 = Bank("Rahul", 1000)

b1.deposit(500)
print(b1.balance)        # 1500

Bank.change_bank("HDFC")
print(Bank.bank_name)    # HDFC
````
