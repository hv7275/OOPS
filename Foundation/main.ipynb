## 🔹 self ko properly samjho

    Galat understanding:

    ❌ self = keyword
    ❌ self = optional

    Sahi understanding:

    ✅ self = current object ka reference

### Example bina self (ERROR):

````python
class Student:
    def set_name(name):
        print(name)

s = Student()
s.set_name("Rahul")  # ERROR
````

### Correct way:

````python
class Student:
    def set_name(self, name):
        self.name = name
s = Student()
s.set_name("Rahul")
print(s.name)

# 🧠 Behind the scenes:
# Student.set_name(s, "Rahul")
# Isliye self compulsory hai.
````

## 🔹 __init__() constructor
    Constructor kya karta hai?

    Object bante hi run hota hai

    Object ko initial state deta hai

### Har object ke liye alag values memory mein store hoti hain.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
s1 = Student("Amit", 22)
s2 = Student("Neha", 21)
````

## 🔹 Instance Variables vs Class Variables
    Instance Variable (object-specific)

````python
class Student:
    def __init__(self, name):
        self.name = name
# Har object ka alag name.
````

### Class Variable (shared)
    🧠 Ek hi copy memory mein hoti hai

````python
class Student:
    college = "IIT"

    def __init__(self, name):
        self.name = name
s1 = Student("Rahul")
s2 = Student("Anita")

print(s1.college)
print(s2.college)
````

    Q3. Class variable vs instance variable?

    ➡️ Shared vs object-specific

````python
class Mobile:
    brand = 'Samsung'
    
    def __init__(self, model, price):
        self.model = model
        self.price = price
    
    def show(self):
        return self.brand, self.model, self.price
    
mobile1 = Mobile('M01', 20000)
print(mobile1.show())
print(mobile1.brand)
print(Mobile.brand)
````

# Why `mobile1.brand` and `Mobile.brand` Give the Same Output

In your code, both `mobile1.brand` and `Mobile.brand` print **`Samsung`**, but they are **not the same thing** conceptually.  
Let’s break it down clearly.
